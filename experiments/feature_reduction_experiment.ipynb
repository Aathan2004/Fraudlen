```python
def create_basic_features(df):

    data = df.copy()

    # --------------------------------------------------------
    # DATE CONVERSION
    # --------------------------------------------------------

    date_columns = [
        "ClaimStartDt",
        "ClaimEndDt",
        "AdmissionDt",
        "DischargeDt",
        "DOB",
        "DOD"
    ]

    for col in date_columns:
        if col in data.columns:
            data[col] = pd.to_datetime(
                data[col],
                errors="coerce"
            )

    # --------------------------------------------------------
    # CLAIM DURATION
    # --------------------------------------------------------

    data["Claim_Days"] = (
        data["ClaimEndDt"]
        - data["ClaimStartDt"]
    ).dt.days

    # --------------------------------------------------------
    # HOSPITAL STAY
    # --------------------------------------------------------

    data["Hospital_Days"] = (
        data["DischargeDt"]
        - data["AdmissionDt"]
    ).dt.days

    # --------------------------------------------------------
    # AGE
    # --------------------------------------------------------

    data["Age"] = (
        data["ClaimStartDt"]
        - data["DOB"]
    ).dt.days / 365.25

    # --------------------------------------------------------
    # REASONABLE VALUE LIMITS
    # --------------------------------------------------------

    data["Claim_Days"] = (
        data["Claim_Days"]
        .clip(lower=0)
    )

    data["Hospital_Days"] = (
        data["Hospital_Days"]
        .clip(lower=0)
    )

    data["Age"] = (
        data["Age"]
        .clip(lower=0, upper=120)
    )

    # --------------------------------------------------------
    # RISK SCORE FROM CHRONIC CONDITIONS
    # --------------------------------------------------------

    chronic_columns = [
        "ChronicCond_Alzheimer",
        "ChronicCond_Heartfailure",
        "ChronicCond_KidneyDisease",
        "ChronicCond_Cancer",
        "ChronicCond_ObstrPulmonary",
        "ChronicCond_Depression",
        "ChronicCond_Diabetes",
        "ChronicCond_IschemicHeart",
        "ChronicCond_Osteoporasis",
        "ChronicCond_rheumatoidarthritis",
        "ChronicCond_stroke"
    ]

    available_chronic = [
        col for col in chronic_columns
        if col in data.columns
    ]

    for col in available_chronic:
        data[col] = pd.to_numeric(
            data[col],
            errors="coerce"
        )

    data["Risk"] = (
        data[available_chronic]
        .replace(2, 0)
        .sum(axis=1)
    )

    return data
```

```python
train_claims = create_basic_features(
    train_claims
)

test_claims = create_basic_features(
    test_claims
)

print("Train:", train_claims.shape)
print("Test :", test_claims.shape)
```

```python
# ============================================================
# PROVIDER-LEVEL FEATURES
# ============================================================

PROVIDER_FEATURES = [
    "count_ClaimID_perProvider",
    "count_ClaimID_perProviderClmProcedureCode_4",
    "count_ClaimID_perProviderClmProcedureCode_3",
    "mean_Hospital_Days_perProvider",
    "mean_DeductibleAmtPaid_perProvider",
    "mean_IPAnnualReimbursementAmt_perProvider",
    "count_ClaimID_perProviderClmProcedureCode_5",
    "mean_Claim_Days_perProvider",
    "mean_InscClaimAmtReimbursed_perProvider",
    "mean_Age_perProvider",
    "mean_OPAnnualDeductibleAmt_perProvider",
    "mean_IPAnnualDeductibleAmt_perProvider",
    "count_ClaimID_perProviderClmDiagnosisCode_10",
    "mean_Risk_perProvider",
    "count_ClaimID_perProviderClmProcedureCode_2",
    "mean_OPAnnualReimbursementAmt_perProvider",
    "count_ClaimID_perProviderAttendingPhysician",
    "count_ClaimID_perProviderClmProcedureCode_1",
    "count_ClaimID_perProviderDiagnosisGroupCode",
    "count_ClaimID_perProviderClmDiagnosisCode_8",
    "count_ClaimID_perProviderBeneID",
    "count_ClaimID_perProviderClmDiagnosisCode_9",
    "count_ClaimID_perProviderClmDiagnosisCode_7",
    "count_ClaimID_perProviderClmDiagnosisCode_4",
    "count_ClaimID_perProviderClmDiagnosisCode_5",
    "count_ClaimID_perProviderOtherPhysician",
    "count_ClaimID_perProviderOperatingPhysician",
    "count_ClaimID_perProviderClmDiagnosisCode_6"
]


def create_provider_features(df):

    data = df.copy()

    provider_features = pd.DataFrame(
        index=data["Provider"].drop_duplicates()
    )

    # --------------------------------------------------------
    # CLAIM COUNT
    # --------------------------------------------------------

    provider_features[
        "count_ClaimID_perProvider"
    ] = data.groupby("Provider")["ClaimID"].transform("count")

    # --------------------------------------------------------
    # PROCEDURE CODE COUNTS
    # --------------------------------------------------------

    for code in [
        "ClmProcedureCode_1",
        "ClmProcedureCode_2",
        "ClmProcedureCode_3",
        "ClmProcedureCode_4",
        "ClmProcedureCode_5"
    ]:

        if code in data.columns:

            provider_features[
                f"count_ClaimID_perProvider{code}"
            ] = data.groupby(
                "Provider"
            )[code].transform(
                lambda x: x.notna().sum()
            )

    # --------------------------------------------------------
    # DIAGNOSIS CODE COUNTS
    # --------------------------------------------------------

    for code in [
        "ClmDiagnosisCode_1",
        "ClmDiagnosisCode_2",
        "ClmDiagnosisCode_3",
        "ClmDiagnosisCode_4",
        "ClmDiagnosisCode_5",
        "ClmDiagnosisCode_6",
        "ClmDiagnosisCode_7",
        "ClmDiagnosisCode_8",
        "ClmDiagnosisCode_9",
        "ClmDiagnosisCode_10"
    ]:

        if code in data.columns:

            provider_features[
                f"count_ClaimID_perProvider{code}"
            ] = data.groupby(
                "Provider"
            )[code].transform(
                lambda x: x.notna().sum()
            )

    # --------------------------------------------------------
    # DIAGNOSIS GROUP CODE
    # --------------------------------------------------------

    if "DiagnosisGroupCode" in data.columns:

        provider_features[
            "count_ClaimID_perProviderDiagnosisGroupCode"
        ] = data.groupby(
            "Provider"
        )["DiagnosisGroupCode"].transform(
            lambda x: x.notna().sum()
        )

    # --------------------------------------------------------
    # BENEFICIARY COUNT
    # --------------------------------------------------------

    provider_features[
        "count_ClaimID_perProviderBeneID"
    ] = data.groupby(
        "Provider"
    )["BeneID"].transform("nunique")

    # --------------------------------------------------------
    # PHYSICIAN COUNTS
    # --------------------------------------------------------

    for physician in [
        "AttendingPhysician",
        "OtherPhysician",
        "OperatingPhysician"
    ]:

        if physician in data.columns:

            provider_features[
                f"count_ClaimID_perProvider{physician}"
            ] = data.groupby(
                "Provider"
            )[physician].transform(
                lambda x: x.notna().sum()
            )

    # --------------------------------------------------------
    # PROVIDER NUMERIC MEANS
    # --------------------------------------------------------

    mean_columns = [
        "Hospital_Days",
        "DeductibleAmtPaid",
        "IPAnnualReimbursementAmt",
        "Claim_Days",
        "InscClaimAmtReimbursed",
        "Age",
        "OPAnnualDeductibleAmt",
        "IPAnnualDeductibleAmt",
        "Risk",
        "OPAnnualReimbursementAmt"
    ]

    for col in mean_columns:

        if col in data.columns:

            provider_features[
                f"mean_{col}_perProvider"
            ] = data.groupby(
                "Provider"
            )[col].transform("mean")

    # --------------------------------------------------------
    # KEEP ONE ROW PER PROVIDER
    # --------------------------------------------------------

    provider_features = (
        provider_features
        .groupby(provider_features.index)
        .first()
    )

    return provider_features.reset_index()
```

```python
# ============================================================
# CREATE ALL SELECTED 62 FEATURES
# CLAIM-LEVEL EXPERIMENT
# ============================================================

import numpy as np
import pandas as pd


# ------------------------------------------------------------
# 1. HELPER: SAFE NUMERIC CONVERSION
# ------------------------------------------------------------

def make_numeric(df, columns):

    data = df.copy()

    for col in columns:

        if col in data.columns:

            data[col] = pd.to_numeric(
                data[col],
                errors="coerce"
            )

    return data


# ------------------------------------------------------------
# 2. CHRONIC CONDITION FEATURES
# ------------------------------------------------------------

CHRONIC_COLUMNS = [
    "ChronicCond_Alzheimer",
    "ChronicCond_Heartfailure",
    "ChronicCond_KidneyDisease",
    "ChronicCond_Cancer",
    "ChronicCond_ObstrPulmonary",
    "ChronicCond_Depression",
    "ChronicCond_Diabetes",
    "ChronicCond_IschemicHeart",
    "ChronicCond_Osteoporasis",
    "ChronicCond_rheumatoidarthritis",
    "ChronicCond_stroke"
]


def create_chronic_features(df):

    data = df.copy()

    for col in CHRONIC_COLUMNS:

        if col in data.columns:

            # Original dataset:
            # 1 = condition present
            # 2 = condition absent
            #
            # Convert to:
            # 1 = Yes
            # 0 = No

            data[col] = pd.to_numeric(
                data[col],
                errors="coerce"
            )

            data[col] = data[col].replace({
                1: 1,
                2: 0
            })

            data[col] = data[col].fillna(0)

    # --------------------------------------------------------
    # Create one-hot columns
    # --------------------------------------------------------

    chronic_data = pd.DataFrame(
        index=data.index
    )

    for col in CHRONIC_COLUMNS:

        if col in data.columns:

            chronic_data[
                f"{col}_0"
            ] = (
                data[col] == 0
            ).astype(int)

            chronic_data[
                f"{col}_1"
            ] = (
                data[col] == 1
            ).astype(int)

    return chronic_data


# ------------------------------------------------------------
# 3. DEMOGRAPHIC FEATURES
# ------------------------------------------------------------

def create_demographic_features(df):

    data = df.copy()

    demographic = pd.DataFrame(
        index=data.index
    )

    # --------------------------------------------------------
    # Gender
    # --------------------------------------------------------

    if "Gender" in data.columns:

        gender = pd.to_numeric(
            data["Gender"],
            errors="coerce"
        )

        demographic["Gender_0"] = (
            gender == 0
        ).astype(int)

        demographic["Gender_1"] = (
            gender == 1
        ).astype(int)

    # --------------------------------------------------------
    # Age category
    # --------------------------------------------------------

    if "Age" in data.columns:

        age = pd.to_numeric(
            data["Age"],
            errors="coerce"
        )

        demographic["Age_Category_Old"] = (
            (age >= 65) &
            (age < 85)
        ).astype(int)

    return demographic


# ------------------------------------------------------------
# 4. GROUP-BASED MEAN FEATURES
# ------------------------------------------------------------

def create_group_mean_features(
    data,
    group_column,
    feature_columns,
    prefix
):

    result = pd.DataFrame(
        index=data.index
    )

    if group_column not in data.columns:
        return result

    for column in feature_columns:

        if column not in data.columns:
            continue

        result[
            f"mean_{column}_per{prefix}"
        ] = (
            data.groupby(
                group_column
            )[column]
            .transform("mean")
        )

    return result


# ------------------------------------------------------------
# 5. PROVIDER FEATURES
# ------------------------------------------------------------

PROVIDER_MEAN_COLUMNS = [
    "Hospital_Days",
    "DeductibleAmtPaid",
    "IPAnnualReimbursementAmt",
    "Claim_Days",
    "InscClaimAmtReimbursed",
    "Age",
    "OPAnnualDeductibleAmt",
    "IPAnnualDeductibleAmt",
    "Risk",
    "OPAnnualReimbursementAmt"
]


PROVIDER_COUNT_COLUMNS = [
    "ClaimID"
]


def create_provider_features_claim_level(data):

    result = pd.DataFrame(
        index=data.index
    )

    # --------------------------------------------------------
    # Claim count per provider
    # --------------------------------------------------------

    result[
        "count_ClaimID_perProvider"
    ] = data.groupby(
        "Provider"
    )["ClaimID"].transform("count")

    # --------------------------------------------------------
    # Procedure counts
    # --------------------------------------------------------

    for col in [
        "ClmProcedureCode_1",
        "ClmProcedureCode_2",
        "ClmProcedureCode_3",
        "ClmProcedureCode_4",
        "ClmProcedureCode_5"
    ]:

        if col in data.columns:

            result[
                f"count_ClaimID_perProvider{col}"
            ] = data.groupby(
                "Provider"
            )[col].transform(
                lambda x: x.notna().sum()
            )

    # --------------------------------------------------------
    # Diagnosis counts
    # --------------------------------------------------------

    for col in [
        "ClmDiagnosisCode_1",
        "ClmDiagnosisCode_2",
        "ClmDiagnosisCode_3",
        "ClmDiagnosisCode_4",
        "ClmDiagnosisCode_5",
        "ClmDiagnosisCode_6",
        "ClmDiagnosisCode_7",
        "ClmDiagnosisCode_8",
        "ClmDiagnosisCode_9",
        "ClmDiagnosisCode_10"
    ]:

        if col in data.columns:

            result[
                f"count_ClaimID_perProvider{col}"
            ] = data.groupby(
                "Provider"
            )[col].transform(
                lambda x: x.notna().sum()
            )

    # --------------------------------------------------------
    # Diagnosis group
    # --------------------------------------------------------

    if "DiagnosisGroupCode" in data.columns:

        result[
            "count_ClaimID_perProviderDiagnosisGroupCode"
        ] = data.groupby(
            "Provider"
        )["DiagnosisGroupCode"].transform(
            lambda x: x.notna().sum()
        )

    # --------------------------------------------------------
    # Beneficiary count
    # --------------------------------------------------------

    result[
        "count_ClaimID_perProviderBeneID"
    ] = data.groupby(
        "Provider"
    )["BeneID"].transform(
        "nunique"
    )

    # --------------------------------------------------------
    # Physician counts
    # --------------------------------------------------------

    for col in [
        "AttendingPhysician",
        "OtherPhysician",
        "OperatingPhysician"
    ]:

        if col in data.columns:

            result[
                f"count_ClaimID_perProvider{col}"
            ] = data.groupby(
                "Provider"
            )[col].transform(
                lambda x: x.notna().sum()
            )

    # --------------------------------------------------------
    # Provider means
    # --------------------------------------------------------

    for col in PROVIDER_MEAN_COLUMNS:

        if col in data.columns:

            result[
                f"mean_{col}_perProvider"
            ] = data.groupby(
                "Provider"
            )[col].transform(
                "mean"
            )

    return result


# ------------------------------------------------------------
# 6. ATTENDING PHYSICIAN FEATURES
# ------------------------------------------------------------

ATTENDING_MEAN_COLUMNS = [
    "DeductibleAmtPaid",
    "IPAnnualDeductibleAmt",
    "InscClaimAmtReimbursed",
    "Hospital_Days",
    "IPAnnualReimbursementAmt",
    "Claim_Days"
]


def create_attending_features(data):

    return create_group_mean_features(
        data=data,
        group_column="AttendingPhysician",
        feature_columns=ATTENDING_MEAN_COLUMNS,
        prefix="AttendingPhysician"
    )


# ------------------------------------------------------------
# 7. BENEFICIARY FEATURE
# ------------------------------------------------------------

def create_beneficiary_features(data):

    result = pd.DataFrame(
        index=data.index
    )

    if "BeneID" in data.columns:

        if "Claim_Days" in data.columns:

            result[
                "mean_Claim_Days_perBeneID"
            ] = data.groupby(
                "BeneID"
            )["Claim_Days"].transform(
                "mean"
            )

    return result


# ------------------------------------------------------------
# 8. DIAGNOSIS / PROCEDURE FEATURES
# ------------------------------------------------------------

def create_diagnosis_procedure_features(data):

    result = pd.DataFrame(
        index=data.index
    )

    # --------------------------------------------------------
    # DiagnosisCode_1
    # --------------------------------------------------------

    if "ClmDiagnosisCode_1" in data.columns:

        result[
            "mean_Hospital_Days_perClmDiagnosisCode_1"
        ] = data.groupby(
            "ClmDiagnosisCode_1"
        )["Hospital_Days"].transform(
            "mean"
        )

    # --------------------------------------------------------
    # ProcedureCode_4
    # --------------------------------------------------------

    if "ClmProcedureCode_4" in data.columns:

        result[
            "mean_Risk_perClmProcedureCode_4"
        ] = data.groupby(
            "ClmProcedureCode_4"
        )["Risk"].transform(
            "mean"
        )

        result[
            "mean_Hospital_Days_perClmProcedureCode_4"
        ] = data.groupby(
            "ClmProcedureCode_4"
        )["Hospital_Days"].transform(
            "mean"
        )

    # --------------------------------------------------------
    # ProcedureCode_5
    # --------------------------------------------------------

    if "ClmProcedureCode_5" in data.columns:

        result[
            "mean_Hospital_Days_perClmProcedureCode_5"
        ] = data.groupby(
            "ClmProcedureCode_5"
        )["Hospital_Days"].transform(
            "mean"
        )

        result[
            "mean_InscClaimAmtReimbursed_perClmProcedureCode_5"
        ] = data.groupby(
            "ClmProcedureCode_5"
        )["InscClaimAmtReimbursed"].transform(
            "mean"
        )

        result[
            "mean_IPAnnualReimbursementAmt_perClmProcedureCode_5"
        ] = data.groupby(
            "ClmProcedureCode_5"
        )["IPAnnualReimbursementAmt"].transform(
            "mean"
        )

        result[
            "mean_OPAnnualDeductibleAmt_perClmProcedureCode_5"
        ] = data.groupby(
            "ClmProcedureCode_5"
        )["OPAnnualDeductibleAmt"].transform(
            "mean"
        )

        result[
            "mean_Claim_Days_perClmProcedureCode_5"
        ] = data.groupby(
            "ClmProcedureCode_5"
        )["Claim_Days"].transform(
            "mean"
        )

    return result


# ------------------------------------------------------------
# 9. CREATE ALL SELECTED FEATURES
# ------------------------------------------------------------

def create_selected_features(data):

    data = data.copy()

    # Basic features were already created,
    # but ensure they exist.

    if "Claim_Days" not in data.columns:

        data["Claim_Days"] = (
            pd.to_datetime(
                data["ClaimEndDt"],
                errors="coerce"
            )
            -
            pd.to_datetime(
                data["ClaimStartDt"],
                errors="coerce"
            )
        ).dt.days

    if "Hospital_Days" not in data.columns:

        data["Hospital_Days"] = (
            pd.to_datetime(
                data["DischargeDt"],
                errors="coerce"
            )
            -
            pd.to_datetime(
                data["AdmissionDt"],
                errors="coerce"
            )
        ).dt.days

    # --------------------------------------------------------
    # Create feature groups
    # --------------------------------------------------------

    chronic_features = create_chronic_features(
        data
    )

    demographic_features = create_demographic_features(
        data
    )

    provider_features = create_provider_features_claim_level(
        data
    )

    attending_features = create_attending_features(
        data
    )

    beneficiary_features = create_beneficiary_features(
        data
    )

    diagnosis_procedure_features = (
        create_diagnosis_procedure_features(
            data
        )
    )

    # --------------------------------------------------------
    # Combine everything
    # --------------------------------------------------------

    feature_data = pd.concat(
        [
            provider_features,
            attending_features,
            beneficiary_features,
            diagnosis_procedure_features,
            chronic_features,
            demographic_features
        ],
        axis=1
    )

    # --------------------------------------------------------
    # Keep ONLY our selected 62 features
    # --------------------------------------------------------

    missing = [
        feature
        for feature in selected_features
        if feature not in feature_data.columns
    ]

    if missing:

        raise ValueError(
            "The following selected features "
            "could not be generated:\n"
            + "\n".join(missing)
        )

    feature_data = feature_data[
        selected_features
    ].copy()

    # --------------------------------------------------------
    # Numeric conversion
    # --------------------------------------------------------

    feature_data = feature_data.apply(
        pd.to_numeric,
        errors="coerce"
    )

    # --------------------------------------------------------
    # Missing values
    # --------------------------------------------------------

    feature_data = feature_data.replace(
        [np.inf, -np.inf],
        np.nan
    )

    feature_data = feature_data.fillna(0)

    return feature_data
```

```python
# ============================================================
# PROVIDER-AWARE 5-FOLD CROSS-VALIDATION
# LOGISTIC REGRESSION
# ============================================================

from sklearn.model_selection import StratifiedGroupKFold
from sklearn.base import clone

cv = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_scores = {
    "accuracy": [],
    "precision": [],
    "recall": [],
    "f1": [],
    "roc_auc": []
}

# Provider ID for every training claim
train_groups = train_claims.loc[
    train_row_mask,
    "Provider"
].reset_index(drop=True)

for fold, (train_idx, val_idx) in enumerate(
    cv.split(
        X_train_final,
        y_train_final,
        groups=train_groups
    ),
    start=1
):

    fold_model = clone(logistic_model)

    X_fold_train = X_train_final.iloc[train_idx]
    X_fold_val = X_train_final.iloc[val_idx]

    y_fold_train = y_train_final.iloc[train_idx]
    y_fold_val = y_train_final.iloc[val_idx]

    fold_model.fit(
        X_fold_train,
        y_fold_train
    )

    probabilities = fold_model.predict_proba(
        X_fold_val
    )[:, 1]

    predictions = (
        probabilities >= 0.50
    ).astype(int)

    cv_scores["accuracy"].append(
        accuracy_score(
            y_fold_val,
            predictions
        )
    )

    cv_scores["precision"].append(
        precision_score(
            y_fold_val,
            predictions
        )
    )

    cv_scores["recall"].append(
        recall_score(
            y_fold_val,
            predictions
        )
    )

    cv_scores["f1"].append(
        f1_score(
            y_fold_val,
            predictions
        )
    )

    cv_scores["roc_auc"].append(
        roc_auc_score(
            y_fold_val,
            probabilities
        )
    )

    print(
        f"Fold {fold}: "
        f"Accuracy={cv_scores['accuracy'][-1]:.4f}, "
        f"Precision={cv_scores['precision'][-1]:.4f}, "
        f"Recall={cv_scores['recall'][-1]:.4f}, "
        f"F1={cv_scores['f1'][-1]:.4f}, "
        f"ROC-AUC={cv_scores['roc_auc'][-1]:.4f}"
    )


print("\n" + "=" * 70)
print("5-FOLD CROSS-VALIDATION RESULTS")
print("=" * 70)

for metric, scores in cv_scores.items():

    print(
        f"{metric.upper():10s}: "
        f"{np.mean(scores):.4f} "
        f"+/- "
        f"{np.std(scores):.4f}"
    )
```

```python
# ============================================================
# CATBOOST THRESHOLD TUNING - VALIDATION ONLY
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

print("=" * 80)
print("CATBOOST THRESHOLD TUNING - VALIDATION SET")
print("=" * 80)

# ------------------------------------------------------------
# 1. Get CatBoost validation probabilities
# ------------------------------------------------------------

catboost_validation_model = validation_models["CatBoost"]

cat_val_probability = (
    catboost_validation_model
    .predict_proba(X_val)[:, 1]
)

print("Validation probabilities generated.")
print("Validation rows:", len(cat_val_probability))


# ------------------------------------------------------------
# 2. Test different thresholds
# ------------------------------------------------------------

cat_threshold_results = []

thresholds = np.arange(
    0.20,
    0.81,
    0.05
)

for threshold in thresholds:

    predictions = (
        cat_val_probability >= threshold
    ).astype(int)

    cat_threshold_results.append({

        "Threshold": threshold,

        "Accuracy": accuracy_score(
            y_val,
            predictions
        ),

        "Precision": precision_score(
            y_val,
            predictions,
            zero_division=0
        ),

        "Recall": recall_score(
            y_val,
            predictions,
            zero_division=0
        ),

        "F1": f1_score(
            y_val,
            predictions,
            zero_division=0
        )
    })


cat_threshold_df = pd.DataFrame(
    cat_threshold_results
)


# ------------------------------------------------------------
# 3. Display threshold results
# ------------------------------------------------------------

print("\n")
print("=" * 80)
print("CATBOOST THRESHOLD RESULTS")
print("=" * 80)

print(
    cat_threshold_df.to_string(
        index=False,
        formatters={
            "Threshold": "{:.2f}".format,
            "Accuracy": "{:.4f}".format,
            "Precision": "{:.4f}".format,
            "Recall": "{:.4f}".format,
            "F1": "{:.4f}".format
        }
    )
)


# ------------------------------------------------------------
# 4. Select threshold with highest validation F1
# ------------------------------------------------------------

best_cat_threshold_row = cat_threshold_df.loc[
    cat_threshold_df["F1"].idxmax()
]

BEST_CAT_THRESHOLD = float(
    best_cat_threshold_row["Threshold"]
)


print("\n")
print("=" * 80)
print("BEST CATBOOST THRESHOLD")
print("=" * 80)

print(
    f"Threshold : {BEST_CAT_THRESHOLD:.2f}"
)

print(
    f"Accuracy  : "
    f"{best_cat_threshold_row['Accuracy']:.4f}"
)

print(
    f"Precision : "
    f"{best_cat_threshold_row['Precision']:.4f}"
)

print(
    f"Recall    : "
    f"{best_cat_threshold_row['Recall']:.4f}"
)

print(
    f"F1        : "
    f"{best_cat_threshold_row['F1']:.4f}"
)
```

```python
# ============================================================
# FIND CATBOOST THRESHOLD
# TARGET:
# Recall >= 90%
# Precision >= 88%
# ============================================================

threshold_search = []

thresholds = np.arange(
    0.05,
    0.51,
    0.01
)

for threshold in thresholds:

    prediction = (
        cat_val_probability >= threshold
    ).astype(int)

    precision = precision_score(
        y_val,
        prediction,
        zero_division=0
    )

    recall = recall_score(
        y_val,
        prediction,
        zero_division=0
    )

    f1 = f1_score(
        y_val,
        prediction,
        zero_division=0
    )

    threshold_search.append({
        "Threshold": threshold,
        "Precision": precision,
        "Recall": recall,
        "F1": f1
    })


threshold_search_df = pd.DataFrame(
    threshold_search
)


# ------------------------------------------------------------
# Find thresholds satisfying both targets
# ------------------------------------------------------------

valid_thresholds = threshold_search_df[
    (threshold_search_df["Recall"] >= 0.90) &
    (threshold_search_df["Precision"] >= 0.88)
].copy()


print("=" * 80)
print("THRESHOLDS MEETING TARGET")
print("=" * 80)

if len(valid_thresholds) == 0:

    print(
        "No threshold satisfies both "
        "Recall >= 90% and Precision >= 88%."
    )

else:

    # Highest F1 among thresholds satisfying targets
    best_target = valid_thresholds.loc[
        valid_thresholds["F1"].idxmax()
    ]

    print(
        valid_thresholds.to_string(
            index=False,
            formatters={
                "Threshold": "{:.2f}".format,
                "Precision": "{:.4f}".format,
                "Recall": "{:.4f}".format,
                "F1": "{:.4f}".format
            }
        )
    )

    print("\n")
    print("=" * 80)
    print("BEST THRESHOLD FOR TARGET")
    print("=" * 80)

    print(
        f"Threshold : {best_target['Threshold']:.2f}"
    )

    print(
        f"Precision : {best_target['Precision']:.4f}"
    )

    print(
        f"Recall    : {best_target['Recall']:.4f}"
    )

    print(
        f"F1        : {best_target['F1']:.4f}"
    )
```
